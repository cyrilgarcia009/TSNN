```python
cs = [0.005, 0.01, 0.03]#, 0.05, 0.1]
effects = ['d_lin', 'd_shift', 'd_cs', 'd_cs_shift', 'd_cond', 'd_all']

import warnings

for c in cs:
    for effect in effects:
        try:
            # print(f'##### {effect} {c} train #####')
            # df = pd.read_parquet(f'sparse_results2/sparse_{effect}_{c}_train.parquet')
            # display(df.iloc[:, 2:].describe().style.background_gradient(axis=1).format(precision=3))

            print(f'##### {effect} {c} test #####')
            df = pd.read_parquet(f'sparse_results3/sparse_{effect}_{c}_test.parquet')
            display(df.iloc[:, 2:].describe().style.background_gradient(axis=1).format(precision=3))
        except:
            continue
```

```python
from scipy import stats

effect = 'd_shift'
c = 0.005
sample1 = pd.read_parquet(f'sparse_results/sparse_{effect}_{c}_test.parquet').full_attention
sample2 = pd.read_parquet(f'sparse_results/sparse_{effect}_{c}_test.parquet').max_sparse
print(sample1.mean(), sample2.mean())
t_statistic, p_value = stats.ttest_ind(sample1, sample2, equal_var=False, alternative='less')
t_statistic, p_value
```

```python
def get_p_values():
    cs = [0.005, 0.01, 0.03]#, 0.05, 0.1]
    effects = [
        # 'd_lin', 
               'd_shift', 'd_cs', 'd_all'
              ]

    import warnings
    res = pd.DataFrame()
    for c in cs:
        for effect in effects:
            try:
                # print(f'##### {effect} {c} train #####')
                # df = pd.read_parquet(f'sparse_results2/sparse_{effect}_{c}_train.parquet')
                # display(df.iloc[:, 2:].describe().style.background_gradient(axis=1).format(precision=3))

                df = pd.read_parquet(f'sparse_results/sparse_{effect}_{c}_test.parquet')
                try:
                    df2 = pd.read_parquet(f'sparse_results3/sparse_{effect}_{c}_test.parquet')
                    df = pd.concat([df, df2])
                except:
                    pass
                sample1 = df.full_attention
                sample2 = df.max_sparse
                t_statistic, p_value = stats.ttest_ind(sample1, sample2, equal_var=False, alternative='less')
                res.loc[c, effect] = p_value
            except:
                continue
    return res
```

```python
get_p_values().style.background_gradient(axis=1).format(precision=3)#.to_latex()
```

```python
def get_avg(effect):
    cs = [0.005, 0.01, 0.03]#, 0.05, 0.1]

    import warnings
    res = []
    for c in cs:
        # for effect in effects:
        try:
            df = pd.read_parquet(f'sparse_results3/sparse_{effect}_{c}_test.parquet')
            try:
                df2 = pd.read_parquet(f'sparse_results/sparse_{effect}_{c}_test.parquet')
                df = pd.concat([df, df2])
            except:
                pass
            
            temp_df = df.mean()
            # temp_df.loc['effect'] = effect
            temp_df.loc['correl'] = str(c)
            
            sample1 = df.full_attention
            sample2 = df.max_sparse
            
            
            
            t_statistic, p_value = stats.ttest_ind(sample1, sample2, equal_var=False, 
                                                   alternative='less'
                                                  )
            
            temp_df.loc['p_value'] = p_value
            
            res.append(temp_df)
        except:
            continue
    return pd.concat(res, axis=1).T.set_index('correl')
```

1.5, 3, 10

```python
to_disp = get_avg('d_shift')[['full_attention', 'max_sparse', 'deterministic_sparse', 'p_value']].T
to_disp.style.background_gradient(axis=0, subset=pd.IndexSlice[['full_attention', 'max_sparse', 'deterministic_sparse'], to_disp.columns],
                                  cmap='coolwarm').format(precision=3)

```

```python
print(to_disp.style.background_gradient(axis=0, subset=pd.IndexSlice[['full_attention', 'max_sparse', 'deterministic_sparse'], to_disp.columns],
                                  cmap='coolwarm').format(precision=3).to_latex(
        column_format='l' + 'c' * len(to_disp.columns),
        hrules=True,
        convert_css=True
    ))
```
